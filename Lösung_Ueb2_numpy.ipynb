```python
a = np.array([1, 2, 3, 4, 5])
b = np.array([10, 20, 30, 40, 50])

print("Addition:", a + b)
print("Subtraktion:", a - b)
print("Multiplikation:", a * b)
print("Mittelwert a:", np.mean(a))
print("Mittelwert b:", np.mean(b))

a1 = a
a1[a1<np.mean(a1)] = 0
print("Alle Elemente <mean(a)=0: ", a1)

b1 = b
b1[b1<np.mean(b1)] = 0
print("Alle Elemente <mean(b)=0: ", b1)
```

## Aufgabe 3 Übungsblatt 2

```python
#!pip install numpy

import numpy as np
import time
```

```python
size = 1000000

x = np.arange(size) # numpy-array zwischen 0-size mit Schrittgröße 1
y = list(range(size)) # Python Liste mit Elementen 0-size mit Schrittgröße 1

print(x[:10])
print(y[:10])

print(sum(x==y)) # Prüfen ob alle Elemente gleich sind
```

```python
# Numpy Array: Quadrieren der Elemente 
start = time.time()
x_sq = np.square(x)
end = time.time()
print('Dauer des Quadrierens: ', end - start)
print(x_sq[:10])
```

```python
# Power
a = np.arange(10)
print('a = ', a)
print('a^2 = ', a^2) # Bitweises XOR
x1 = 3 # Binär '11'
x2 = 2 # Binär '10'
print('x1^x2 = ', x1^x2) # Bitweises XOR: 1 XOR 1 = 0, 1 XOR 0 = 1, Ergebnis ist '01' in dezimalschreibeweise 1

print('square(a) = ', np.square(a))
print('a*a = ', a*a)
print('a**2 = ', a**2)

```

```python
# List - Loop - Quadrieren der Elemente
y_sq = y
start = time.time()
for i in range(len(y)):
    y_sq[i] = y[i]**2
end = time.time()
print(end - start)

print(y_sq[:10])
print(sum(y_sq == x_sq))
```

```python
# List - Loop - List Comprehensions
start = time.time()
y_squared = [i ** 2 for i in y]
end = time.time()
print(end - start)

print(y_sq[:10])
print(sum(y_sq == x_sq))
```
